# Sectional Curvature on SU(3)

---

## Exercise
The sectional curvature of the bi-invariant metric is $K(X,Y)=\tfrac14\|[X,Y]\|^2$, with $g(A,B)=-\mathrm{Tr}(AB)$ for skew-Hermitian $A,B$. With $E_{jk}=|j\rangle\langle k|$ on $\mathbb{C}^3$:
$X_1=\tfrac{-i}{\sqrt2}\mathrm{diag}(1,-1,0)$, $Y_1=\tfrac{-i}{\sqrt6}\mathrm{diag}(1,1,-2)$, $X_2=\tfrac{-i}{\sqrt2}(E_{12}+E_{21})$, $Y_2=\tfrac{1}{\sqrt2}(E_{21}-E_{12})$, $Y_3=\tfrac{-i}{\sqrt2}(E_{23}+E_{32})$.

**(a)** Show each generator is in $\mathfrak{su}(3)$ and that $(X_1,Y_1)$, $(X_2,Y_2)$, $(X_2,Y_3)$ are orthonormal pairs.
**(b)** Compute $[X,Y]$ and $K(X,Y)$ for the three pairs.
**(c)** For which pair do two evolutions leaving the identity along $X$ and $X+\epsilon Y$ stay parallel, and for which do they reconverge fastest?

## Solution
**(a)** All five are skew-Hermitian and traceless; the norms are 1, and the cross terms vanish ($\mathrm{Tr}(X_1Y_1)\propto 1-1=0$, $X_2Y_2=-\tfrac{i}{2}(E_{11}-E_{22})$ is traceless, $X_2Y_3=-\tfrac12E_{13}$ has no diagonal).

**(b)** $[X_1,Y_1]=0$, $K=0$. $[X_2,Y_2]=-i\,\mathrm{diag}(1,-1,0)$, $\|[X_2,Y_2]\|^2=2$, $K=\tfrac12$ (an $\mathfrak{su}(2)$ block, as for $SU(2)$). $[X_2,Y_3]=\tfrac12(E_{31}-E_{13})$, $\|[X_2,Y_3]\|^2=\tfrac12$, $K=\tfrac18$.

**(c)** The diagonal pair commutes: a flat torus, parallel geodesics. The same-block pair reconverges fastest. The pair sharing one state has a quarter of that curvature.

---
## Numerical verification

```python
import numpy as np
g = lambda A, B: np.real(np.trace(A.conj().T @ B))   # Hilbert-Schmidt metric, = -Tr(AB) on skew-Hermitian
comm = lambda A, B: A @ B - B @ A
E = lambda i, j: np.outer(np.eye(3)[i], np.eye(3)[j])
X1 = -1j*np.diag([1, -1, 0])/np.sqrt(2);  Y1 = -1j*np.diag([1, 1, -2])/np.sqrt(6)
X2 = -1j*(E(0,1) + E(1,0))/np.sqrt(2);    Y2 = (E(1,0) - E(0,1))/np.sqrt(2)
Y3 = -1j*(E(1,2) + E(2,1))/np.sqrt(2)
for name, (X, Y) in {"(X1,Y1)": (X1, Y1), "(X2,Y2)": (X2, Y2), "(X2,Y3)": (X2, Y3)}.items():
    assert np.allclose(X.conj().T, -X) and np.allclose(Y.conj().T, -Y)          # skew-Hermitian
    assert np.isclose(np.trace(X), 0) and np.isclose(np.trace(Y), 0)            # traceless
    assert np.isclose(g(X, X), 1) and np.isclose(g(Y, Y), 1) and np.isclose(g(X, Y), 0)
    C = comm(X, Y); K = g(C, C)/4
    print(f"{name}: |[X,Y]|^2 = {g(C,C):.4f},  K = {K:.4f}")
assert np.isclose(g(comm(X1,Y1), comm(X1,Y1))/4, 0)
assert np.isclose(g(comm(X2,Y2), comm(X2,Y2))/4, 1/2)
assert np.isclose(g(comm(X2,Y3), comm(X2,Y3))/4, 1/8)
print("K = 0, 1/2, 1/8 VERIFIED")
```

## Takeaway

The curvature of a plane in $\mathfrak{su}(3)$ measures how much the two generators overlap in the states they couple: $K=0$ for commuting generators, $\tfrac12$ within one $\mathfrak{su}(2)$ block, $\tfrac18$ for blocks sharing one state.
