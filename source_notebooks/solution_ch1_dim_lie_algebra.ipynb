# Unitaries That Commute with a Hamiltonian

---

## Exercise
**(a)** If $\hat H$ has $D$ distinct eigenvalues, show that the unitaries commuting with $\hat H$ are those diagonal in its eigenbasis, a torus of dimension $D$.
**(b)** With degeneracies $d_1,\dots,d_m$, show they form $U(d_1)\times\cdots\times U(d_m)$, of dimension $\sum_k d_k^2$.
**(c)** For $D=4$ evaluate the patterns $(1,1,1,1),(2,1,1),(2,2),(3,1),(4)$ and compare with $\dim U(4)=16$. What fraction of $U(2^N)$ does the non-degenerate family fill?

## Solution
**(a)** $[\hat U,\hat H]=0$ means $\hat U$ maps each eigenspace into itself; a one-dimensional eigenspace only by a phase, so $\hat U$ is diagonal with $D$ free phases.
**(b)** On a $d_k$-dimensional eigenspace $\hat U$ may act as any element of $U(d_k)$, of dimension $d_k^2$.
**(c)** $4, 6, 8, 10, 16$. For $N$ qubits: $2^N$ of $4^N$, a fraction $2^{-N}$.

---
## Numerical verification

```python
import numpy as np
def commutant_dim(eigs):
    """Real dimension of {X skew-Hermitian : [X, H] = 0} for H = diag(eigs)."""
    D = len(eigs); H = np.diag(eigs).astype(complex); basis = []
    for i in range(D):
        A = np.zeros((D, D), complex); A[i, i] = 1j; basis.append(A)
        for j in range(i+1, D):
            A = np.zeros((D, D), complex); A[i, j], A[j, i] = 1, -1; basis.append(A)
            A = np.zeros((D, D), complex); A[i, j], A[j, i] = 1j, 1j; basis.append(A)
    M = np.array([(B @ H - H @ B).ravel() for B in basis]).T
    return len(basis) - np.linalg.matrix_rank(np.vstack([M.real, M.imag]))
for pattern in [(1,1,1,1), (2,1,1), (2,2), (3,1), (4,)]:
    eigs = sum([[k]*m for k, m in enumerate(pattern)], [])
    d = commutant_dim(np.array(eigs, float))
    print(pattern, "dim =", d, " sum d_k^2 =", sum(m*m for m in pattern))
    assert d == sum(m*m for m in pattern)
for N in range(1, 6):
    print(f"N={N}: non-degenerate family dimension 2^N = {2**N} of 4^N = {4**N}, fraction {2**N/4**N:.4f}")
print("sum d_k^2 VERIFIED")
```

## Takeaway

A Hamiltonian evolution explores only the commutant of $\hat H$, of dimension $\sum_k d_k^2$, which for a non-degenerate spectrum is an exponentially small slice of $U(2^N)$.
